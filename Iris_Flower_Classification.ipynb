```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import joblib

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("Python Version:", sys.version)
print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)
print("Scikit-Learn Version:", sklearn.__version__)

print("\nUpload the iris.data file:")

uploaded = files.upload()

print("\nUploaded Files:")

for filename in uploaded.keys():
    print(filename)

dataset_filename = "iris.data"

if dataset_filename not in uploaded:
    raise FileNotFoundError(
        "iris.data was not uploaded. Please upload the UCI iris.data file."
    )

dataset_path = os.path.join(
    "/content",
    dataset_filename
)

print("\nDataset found successfully.")
print("Dataset path:", dataset_path)

column_names = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm",
    "Species"
]

df = pd.read_csv(
    dataset_path,
    header=None,
    names=column_names
)

df = df.dropna(how="all").reset_index(drop=True)

df["Species"] = df["Species"].astype(str).str.strip()

print("\nDataset loaded successfully.")
print("Dataset Shape:", df.shape)

if df.shape[1] != 5:
    raise ValueError(
        "The uploaded iris.data file does not have the expected 5 columns."
    )

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nData Types:")
display(df.dtypes)

missing_values = df.isnull().sum()

print("\nMissing Values:")
display(missing_values)

total_missing_values = df.isnull().sum().sum()

print("\nTotal Missing Values:")
print(total_missing_values)

duplicate_count = df.duplicated().sum()

print("\nNumber of Duplicate Rows:")
print(duplicate_count)

print("\nUnique Species:")
print(df["Species"].unique())

print("\nNumber of Classes:")
print(df["Species"].nunique())

print("\nClass Distribution:")
display(df["Species"].value_counts())

numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

if df[numeric_columns].isnull().sum().sum() > 0:
    raise ValueError(
        "Missing values were found in the numerical features."
    )

if df["Species"].isnull().sum() > 0:
    raise ValueError(
        "Missing values were found in the target variable."
    )

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Species",
    hue="Species",
    palette="Set2",
    legend=False
)

plt.title("Iris Species Distribution")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()

sns.pairplot(
    df,
    hue="Species",
    palette="Set2",
    diag_kind="hist"
)

plt.suptitle(
    "Pairplot of Iris Dataset",
    y=1.02
)

plt.show()

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    square=True
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(
    2,
    2,
    figsize=(12, 8)
)

for ax, feature in zip(
    axes.flatten(),
    numeric_columns
):
    sns.histplot(
        data=df,
        x=feature,
        hue="Species",
        kde=True,
        palette="Set2",
        ax=ax
    )

    ax.set_title(f"{feature} Distribution")

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(
    2,
    2,
    figsize=(12, 8)
)

for ax, feature in zip(
    axes.flatten(),
    numeric_columns
):
    sns.boxplot(
        data=df,
        x="Species",
        y=feature,
        hue="Species",
        palette="Set2",
        legend=False,
        ax=ax
    )

    ax.set_title(f"{feature} by Species")
    ax.tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()

X = df[numeric_columns]

y = df["Species"]

print("\nFeature Matrix Shape:")
print(X.shape)

print("\nTarget Shape:")
print(y.shape)

print("\nSelected Features:")
print(list(X.columns))

print("\nTarget Variable:")
print(y.name)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTrain-Test Split:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nTraining Target Distribution:")
display(y_train.value_counts())

print("\nTesting Target Distribution:")
display(y_test.value_counts())

model_pipeline = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        LogisticRegression(
            max_iter=200,
            random_state=42
        )
    )
])

print("\nModel Pipeline:")
display(model_pipeline)

print("\nTraining Logistic Regression Model...")

model_pipeline.fit(
    X_train,
    y_train
)

print("Model Training Completed Successfully.")

y_pred = model_pipeline.predict(X_test)

print("\nPredictions Generated Successfully.")

print("\nPredicted Classes:")
print(y_pred)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nModel Evaluation:")
print(f"Accuracy Score: {accuracy:.4f}")
print(f"Accuracy Percentage: {accuracy * 100:.2f}%")

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)

report = classification_report(
    y_test,
    y_pred
)

print("\nClassification Report:")
print(report)

class_names = model_pipeline.named_steps[
    "model"
].classes_

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.tight_layout()
plt.show()

prediction_results = pd.DataFrame({
    "Actual Species": y_test.values,
    "Predicted Species": y_pred
})

prediction_results["Correct"] = (
    prediction_results["Actual Species"]
    ==
    prediction_results["Predicted Species"]
)

print("\nActual vs Predicted Results:")
display(prediction_results)

incorrect_predictions = prediction_results[
    prediction_results["Correct"] == False
]

print("\nNumber of Incorrect Predictions:")
print(len(incorrect_predictions))

if len(incorrect_predictions) > 0:
    print("\nIncorrect Predictions:")
    display(incorrect_predictions)
else:
    print("The model correctly classified all test samples.")

model_filename = "iris_logistic_regression_model.pkl"

joblib.dump(
    model_pipeline,
    model_filename
)

print("\nModel saved successfully:")
print(model_filename)

loaded_model = joblib.load(
    model_filename
)

loaded_predictions = loaded_model.predict(
    X_test
)

loaded_accuracy = accuracy_score(
    y_test,
    loaded_predictions
)

print("\nSaved Model Verification:")
print(
    f"Accuracy of Loaded Model: "
    f"{loaded_accuracy * 100:.2f}%"
)

print("\n" + "=" * 60)
print("FINAL MODEL RESULTS")
print("=" * 60)

print(f"Dataset Samples: {len(df)}")
print(f"Number of Features: {X.shape[1]}")
print(f"Training Samples: {len(X_train)}")
print(f"Testing Samples: {len(X_test)}")
print("Model: Logistic Regression")
print("Preprocessing: StandardScaler")
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"Incorrect Predictions: {len(incorrect_predictions)}")
print(f"Saved Model: {model_filename}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)

print("=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)
```
